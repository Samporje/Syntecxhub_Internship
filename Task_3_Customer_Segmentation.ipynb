```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
```

```python

df = pd.read_csv("Mall_Customers.csv")

print("Dataset Preview:")
print(df.head())
```

```python

X = df[['Age', 'Spending Score (1-100)']]
```

```python
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

```python
wcss = []

for k in range(1, 11):
    kmeans = KMeans(n_clusters=k, init='k-means++', random_state=42)
    kmeans.fit(X_scaled)
    wcss.append(kmeans.inertia_)

plt.figure(figsize=(8,5))
plt.plot(range(1, 11), wcss, marker='o')
plt.title("Elbow Method")
plt.xlabel("Number of Clusters (k)")
plt.ylabel("WCSS")
plt.show()
```

```python
kmeans = KMeans(n_clusters=5, init='k-means++', random_state=42)
df['Cluster'] = kmeans.fit_predict(X_scaled)
```

```python
plt.figure(figsize=(8,6))
sns.scatterplot(
    x=df['Age'],
    y=df['Spending Score (1-100)'],
    hue=df['Cluster'],
    palette='Set2'
)
plt.title("Customer Segmentation")
plt.xlabel("Age")
plt.ylabel("Spending Score")
plt.show()
```

```python
profile = df.groupby('Cluster')[['Age', 'Spending Score (1-100)']].mean()
print("\nCluster Profiles:")
print(profile)
```

```python
print("\nMarketing Strategy per Segment:\n")

for cluster in profile.index:
    age = profile.loc[cluster, 'Age']
    spend = profile.loc[cluster, 'Spending Score (1-100)']

    print(f"Cluster {cluster}:")
    print(f"Average Age: {age:.1f}")
    print(f"Average Spending Score: {spend:.1f}")

    if spend > 70:
        print("→ High-value customers: Premium offers & loyalty programs")
    elif spend > 40:
        print("→ Medium spenders: Discounts & personalized promotions")
    else:
        print("→ Low spenders: Awareness campaigns & budget offers")
    
    print("-" * 50)
```

```python
df.to_csv("Customer_Segments.csv", index=False)
print("\nClustered data saved as Customer_Segments.csv")
```
